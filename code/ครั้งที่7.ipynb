```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel('Spotify.xlsx')
df.columns = df.columns.str.strip()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Pie Chart: Top 5 Primary Genres + Others
genre_counts = df['Primary Genre'].value_counts()
top_genres = genre_counts.head(5)
others = pd.Series({'Others': genre_counts[5:].sum()})
pie_data = pd.concat([top_genres, others])

axes[0].pie(pie_data, labels=pie_data.index, autopct='%1.1f%%', startangle=140)
axes[0].set_title('Proportion of Primary Genres')

# 2. Scatter Plot: Lead Streams vs Feature Streams
axes[1].scatter(df['Lead Streams (in millions)'], df['Feature Streams (in millions)'], alpha=0.7, color='teal')
axes[1].set_xlabel('Lead Streams (in millions)')
axes[1].set_ylabel('Feature Streams (in millions)')
axes[1].set_title('Lead Streams vs Feature Streams')

# 3. Line Chart: Total Streams by Debut Year (Sum or Mean)
yearly_streams = df.groupby('Debut Year')['Total Streams (in millions)'].sum()
axes[2].plot(yearly_streams.index, yearly_streams.values, marker='o', color='crimson')
axes[2].set_xlabel('Debut Year')
axes[2].set_ylabel('Total Streams (in millions)')
axes[2].set_title('Total Streams by Debut Year')

plt.tight_layout()
plt.savefig('charts.png')
print("Successfully created charts!")
#กราฟวงกลม (Pie Chart): แสดงสัดส่วนของแนวเพลงหลัก (Primary Genre) โดยจัดกลุ่มแนวเพลงอันดับ 1–5 และรวมส่วนที่เหลือเป็นหมวดหมู่ "Others"

#กราฟการกระจาย (Scatter Plot): เปรียบเทียบความสัมพันธ์ระหว่างยอดสตรีมในฐานะนักร้องหลัก (Lead Streams) กับยอดสตรีมในฐานะศิลปินรับเชิญ (Feature Streams)

#กราฟเส้น (Line Chart): แสดงแนวโน้มผลรวมยอดสตรีมทั้งหมด (Total Streams) จำแนกตามปีที่เริ่มเปิดตัวผลงาน (Debut Year)
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. โหลดข้อมูลจากไฟล์ Spotify
df = pd.read_csv("C:/Users/User/Desktop/Spotify.csv")

# 2. ดึงข้อมูล 4 แถวแรกจากไฟล์ตามที่คุณต้องการ
# ดึง Lead Streams มาทำเป็น X และดึง Total Streams มาทำเป็น y
x_data = df['Lead Streams (in millions)'].head(4).values  
y = df['Total Streams (in millions)'].head(4).values

# 3. ใช้การ Reshape แบบที่คุณเขียน เพื่อเปลี่ยน X ให้เป็น 2D Array ขนาด (4, 1)
X = x_data.reshape(-1, 1)

# 4. สร้างและเทรนโมเดลหาเส้นตรงที่ตรงที่สุด (Linear Regression)
model = LinearRegression()
model.fit(X, y)

# 5. พยากรณ์ผลลัพธ์
y_pred = model.predict(X)

# 6. คำนวณค่าสถิติวัดผล
r2 = r2_score(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

# แสดงผลลัพธ์
print(f"ขนาดของ X (Feature Shape): {X.shape}")
print(f"ค่า X ข้างใน:\n{X}\n")
print(f"สมการเส้นตรง: Y = ({model.coef_[0]:.4f} * X) + ({model.intercept_:.4f})")
print(f"R-squared (R²): {r2:.4f}")
print(f"RMSE: {rmse:.4f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. โหลดข้อมูลจากไฟล์ Spotify ของคุณ
df = pd.read_csv("C:/Users/User/Desktop/Spotify.csv")

# 2. ดึงข้อมูล 4 แถวแรก: Debut Year เป็น X และ Total Streams เป็น y
x_data = df['Debut Year'].head(4).values  
y = df['Total Streams (in millions)'].head(4).values

# 3. Reshape ตัวแปร X ให้เป็น 2D Array ตามรูปแบบที่คุณดึงมาใช้
X = x_data.reshape(-1, 1)

# 4. สร้างและเทรนโมเดล Linear Regression
model = LinearRegression()
model.fit(X, y)

# 5. ดึงค่าสถิติสำคัญออกมา
slope = model.coef_[0]         # ค่าความชัน
intercept = model.intercept_   # ค่าจุดตัดแกน Y

print("=== ผลลัพธ์ตัวเลขทางสถิติ ===")
print(f"ค่าความชัน (Slope): {slope:.4f}")
print(f"ค่าจุดตัดแกน Y (Intercept): {intercept:.4f}")
print(f"สมการคือ: Total Streams = ({slope:.4f} * Debut Year) + {intercept:.4f}\n")

# 6. สร้างกราฟแสดงผล
plt.figure(figsize=(8, 6))

# วาดจุดข้อมูลจริง (Scatter Plot) ของศิลปินทั้ง 4 คน
plt.scatter(X, y, color='blue', s=100, label='Actual Data (4 Artists)')

# วาดเส้นตรงที่ตรงที่สุด (Regression Line)
y_pred = model.predict(X)
plt.plot(X, y_pred, color='red', linewidth=2, label='Best Fit Line')

# ใส่ชื่อกราฟและแกน
plt.title('Total Streams vs Debut Year (Top 4 Artists)')
plt.xlabel('Debut Year')
plt.ylabel('Total Streams (in millions)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

# แสดงกราฟขึ้นมาบนหน้าจอ
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. โหลดข้อมูลจากไฟล์ Spotify ของคุณ
df = pd.read_csv("C:/Users/User/Desktop/Spotify.csv")

# 2. ดึงข้อมูล 4 แถวแรก: Debut Year เป็น X และ Total Streams เป็น y
x_data = df['Debut Year'].head(4).values  
y = df['Total Streams (in millions)'].head(4).values

# 3. Reshape ตัวแปร X ให้เป็น 2D Array
X = x_data.reshape(-1, 1)

# 4. สร้างและเทรนโมเดล Linear Regression
model = LinearRegression()
model.fit(X, y)

# 5. ดึงค่าสถิติสำคัญรวมถึง R-squared (ค่าสถิติเปรียบเทียบความสัมพันธ์)
slope = model.coef_[0]         # ค่าความชัน
intercept = model.intercept_   # ค่าจุดตัดแกน Y
r_squared = model.score(X, y)  # ค่า R-squared (ความแม่นยำ/การอธิบายโมเดล)

print("================ ผลลัพธ์ตัวเลขทางสถิติ ================")
print(f"ค่าความชัน (Slope): {slope:.4f}")
print(f"ค่าจุดตัดแกน Y (Intercept): {intercept:.4f}")
print(f"ค่าอาร์สแควร์ (R-squared): {r_squared:.4f} (หรือ {r_squared*100:.2f}%)")
print(f"สมการคือ: Total Streams = ({slope:.4f} * Debut Year) + {intercept:.4f}")
print("==================================================\n")

# 6. สร้างกราฟแสดงผล
plt.figure(figsize=(8, 6))
plt.scatter(X, y, color='blue', s=100, label='Actual Data (4 Artists)')

y_pred = model.predict(X)
plt.plot(X, y_pred, color='red', linewidth=2, label=f'Best Fit Line ($R^2$: {r_squared:.2f})')

plt.title('Total Streams vs Debut Year (Top 4 Artists)')
plt.xlabel('Debut Year')
plt.ylabel('Total Streams (in millions)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. โหลดข้อมูลจากไฟล์ Spotify
df = pd.read_csv("C:/Users/User/Desktop/Spotify.csv")

# 2. ดึงข้อมูล 4 แถวแรก
# สมมติว่ามีคอลัมน์ชื่อ 'Debut Year', 'Total Streams (in millions)' และ 'Lead Streams (in millions)'
# (หากในไฟล์จริงชื่อคอลัมน์ต่างออกไปเล็กน้อย สามารถปรับเปลี่ยนในเครื่องหมาย ' ' ได้ครับ)
years = df['Debut Year'].head(4).astype(str).values  # แปลงปีเป็นข้อความเพื่อให้แกน X เรียงตามลำดับข้อมูล
total_streams = df['Total Streams (in millions)'].head(4).values
lead_streams = df['Lead Streams (in millions)'].head(4).values

# 3. สร้างกราฟเส้นเปรียบเทียบข้อมูล
plt.figure(figsize=(10, 6))

# เส้นที่ 1: Total Streams (สีเขียวเข้ม)
plt.plot(years, total_streams, color='#1DB954', marker='o', linewidth=2.5, markersize=8, label='Total Streams')

# เส้นที่ 2: Lead Streams (สีน้ำเงิน/ฟ้า)
plt.plot(years, lead_streams, color='#1976D2', marker='s', linewidth=2.5, markersize=8, label='Lead Streams')

# 4. ใส่ตัวเลขกำกับบนจุดของทั้งสองเส้นเพื่อการเปรียบเทียบที่ชัดเจน
for i in range(len(years)):
    # ตัวเลขสำหรับ Total Streams (อยู่ด้านบนจุด)
    plt.text(years[i], total_streams[i] + (max(total_streams)*0.02), f"{total_streams[i]:,.0f}M", 
             ha='center', va='bottom', fontsize=9, weight='bold', color='#1DB954')
    # ตัวเลขสำหรับ Lead Streams (อยู่ด้านล่างจุด)
    plt.text(years[i], lead_streams[i] - (max(total_streams)*0.04), f"{lead_streams[i]:,.0f}M", 
             ha='center', va='top', fontsize=9, weight='bold', color='#1976D2')

# 5. ตกแต่งรายละเอียดกราฟ
plt.title('Comparison: Total Streams vs Lead Streams (Top 4 Artists)', fontsize=14, pad=15, weight='bold')
plt.xlabel('Debut Year (ปีที่เดบิวต์)', fontsize=12, labelpad=10)
plt.ylabel('Streams (in millions)', fontsize=12, labelpad=10)

# เปิดเส้นตารางและแสดงคำอธิบายเส้น (Legend)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=11, loc='upper right')

# แสดงกราฟ
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. โหลดข้อมูลจากไฟล์ Spotify
df = pd.read_csv("C:/Users/User/Desktop/Spotify.csv")

# 2. สร้างกราฟ Scatter Plot เปรียบเทียบความสัมพันธ์ของทั้ง 2 ตัวแปร
plt.figure(figsize=(9, 7))

# พล็อตข้อมูลศิลปินทั้งหมดในไฟล์ โดยให้แกน X ชนกับแกน Y
sns.scatterplot(data=df, x='Lead Streams (in millions)', y='Total Streams (in millions)', 
                color='#1DB954', s=80, alpha=0.7, edgecolors='black')

# 3. วาดเส้นทแยงมุม 45 องศา (Reference Line) เพื่อเปรียบเทียบสัดส่วน
# ยอดสตรีมรวมไม่มีทางน้อยกว่ายอดสตรีมเพลงหลัก ดังนั้นจุดข้อมูลทั้งหมดจะอยู่เหนือเส้นนี้เสมอ
max_val = max(df['Total Streams (in millions)'].max(), df['Lead Streams (in millions)'].max())
plt.plot([0, max_val], [0, max_val], color='red', linestyle='--', alpha=0.7, label='1:1 Ratio (Lead = Total)')

# 4. ตกแต่งรายละเอียดกราฟ
plt.title('Correlation: Total Streams vs Lead Streams', fontsize=14, pad=15, weight='bold')
plt.xlabel('Lead Streams (in millions)', fontsize=12, labelpad=10)
plt.ylabel('Total Streams (in millions)', fontsize=12, labelpad=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. โหลดข้อมูลจากไฟล์ Spotify
df = pd.read_csv("C:/Users/User/Desktop/Spotify.csv")

# 2. ดึงข้อมูล 4 แถวแรก: ให้ Lead Streams เป็น X และ Total Streams เป็น y
x_data = df['Lead Streams (in millions)'].head(4).values  
y_actual = df['Total Streams (in millions)'].head(4).values

# 3. Reshape ตัวแปร X ให้เป็น 2D Array ตามเงื่อนไขของ scikit-learn
X = x_data.reshape(-1, 1)

# 4. สร้างและเทรนโมเดล Linear Regression
model = LinearRegression()
model.fit(X, y_actual)

# 5. ดึงค่าสถิติสำคัญออกมาเพื่อดูหน้าตาสมการ
slope = model.coef_[0]
intercept = model.intercept_

# 6. คำนวณหาค่า y_hat (ค่าที่ได้จากการพยากรณ์ตามเส้น Regression)
y_hat = model.predict(X)

# 7. คำนวณค่าคลาดเคลื่อน (Residuals หรือ Error = y_actual - y_hat)
residuals = y_actual - y_hat

# แสดงผลลัพธ์สมการและตารางเปรียบเทียบ
print("=== สมการ Linear Regression ===")
print(f"Total Streams (y_hat) = ({slope:.4f} * Lead Streams) + {intercept:.4f}\n")

print("=== ตารางเปรียบเทียบค่าจริง (y) กับ ค่าพยากรณ์ (y_hat) ===")
comparison_df = pd.DataFrame({
    'Lead Streams (X)': x_data,
    'Actual Total (y)': y_actual,
    'Predicted Total (y_hat)': y_hat,
    'Error (Residual)': residuals
})

# ปรับรูปแบบการแสดงผลทศนิยมให้ดูง่าย
print(comparison_df.to_string(index=False, formatters={
    'Lead Streams (X)': '{:,.2f}'.format,
    'Actual Total (y)': '{:,.2f}'.format,
    'Predicted Total (y_hat)': '{:,.2f}'.format,
    'Error (Residual)': '{:,.2f}'.format
}))

```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. โหลดข้อมูลจากไฟล์ Spotify
df = pd.read_csv("C:/Users/User/Desktop/Spotify.csv")

# 2. ดึงข้อมูล 4 แถวแรก
# X = Solo Streams (ตัวแปรต้น), y = % of Solo Streams (ตัวแปรตาม)
x_data = df['Solo Streams (in millions)'].head(4).values  
y = df['% of Solo Streams'].head(4).values

# 3. Reshape X ให้เป็น 2D Array ตามเงื่อนไขของ Scikit-Learn
X = x_data.reshape(-1, 1)

# 4. สร้างและเทรนโมเดล Linear Regression
model = LinearRegression()
model.fit(X, y)

# 5. ดึงค่าสถิติสำคัญออกมา
slope = model.coef_[0]         # ค่าความชัน
intercept = model.intercept_   # ค่าจุดตัดแกน Y
r_squared = model.score(X, y)  # ค่า R-squared

print("================ ผลลัพธ์ตัวเลขทางสถิติ ================")
print(f"ค่าความชัน (Slope): {slope:.6f}")
print(f"ค่าจุดตัดแกน Y (Intercept): {intercept:.4f}")
print(f"ค่าอาร์สแควร์ (R-squared): {r_squared:.4f} (หรือ {r_squared*100:.2f}%)")
print(f"สมการคือ: % of Solo Streams = ({slope:.6f} * Solo Streams) + {intercept:.4f}")
print("==================================================\n")

# 6. พล๊อตกราฟเปรียบเทียบ (Scatter + Regression Line)
plt.figure(figsize=(9, 6))
plt.scatter(X, y, color='#E91E63', s=100, zorder=3, edgecolors='black', label='Actual Data (4 Artists)')

# วาดเส้น Best Fit Line
y_pred = model.predict(X)
plt.plot(X, y_pred, color='#2196F3', linewidth=2, label=f'Best Fit Line ($R^2$: {r_squared:.2f})')

plt.title('Comparison: % of Solo Streams vs Solo Streams', fontsize=14, pad=15, weight='bold')
plt.xlabel('Solo Streams (in millions)', fontsize=12)
plt.ylabel('% of Solo Streams', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
